```python
%matplotlib widget

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider
x = np.linspace(-6,6,100)
y = np.exp(-x**2)

fig = plt.figure(figsize=(10,10))
ax = fig.add_axes([0.25,0.5,0.5,0.5])
line, = ax.plot(x,y)
sigma_slider = Slider(plt.axes([0.25,0.4,0.5,0.02]),label='sigma', valmin=1, valmax=100,valinit=1.0)

def update(value):
    sigma = sigma_slider.val
    new_y = np.exp(-sigma*(x**2))
    line.set_ydata(new_y)
    fig.canvas.draw_idle()
    


sigma_slider.on_changed(update)

plt.show()
```
