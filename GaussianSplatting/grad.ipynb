```python
import torch


it = 10000

def Render(x,y):
    return x**2+y**2

device = 'cuda'
target_x = torch.tensor(0.0, device=device, dtype=torch.float32)
target_y = torch.tensor(0.0, device=device, dtype=torch.float32)

init_x = torch.tensor(11.0, device=device, dtype=torch.float32)
init_y = torch.tensor(13.0, device=device, dtype=torch.float32)

x = init_x.clone().requires_grad_(True)
y = init_y.clone().requires_grad_(True)

target = Render(target_x, target_y)
optimizer = torch.optim.Adam([x,y], lr=0.05)

step = 100
for i in range(it):
    optimizer.zero_grad()

    image = Render(x,y)
    loss = torch.mean((target - image)**2)

    loss.backward()
    optimizer.step()

    if i % step == 0:
        print(f"{i} - x : {x.item()}, y : {y.item()}")

```
